```python
import cv2
import os

video_path = r"C:\Users\Admin\Downloads\2025-04-01_163954_VID001.mp4"
output_dir = r"D:\Khanh\CardioVis\frameslabel"
os.makedirs(output_dir, exist_ok=True)

cap = cv2.VideoCapture(video_path, cv2.CAP_ANY)

count = 0
while True:
    ret, frame = cap.read()
    if not ret:
        print("Stopped at:", count)
        break
    out_path = os.path.join(output_dir, f"frame_{count:05d}.jpg")
    cv2.imwrite(out_path, frame)
    count += 1

cap.release()
print(f"Đã trích {count} frame → {output_dir}")
```
